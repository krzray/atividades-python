questão 5

```python
nota1 = float(input("Digite sua primeira nota: "))
nota2 = float(input("Digite sua segunda nota: "))

media = (nota1 + nota2)/ 2

if media >= 7:
  print("Aprovado")
elif 5 <= media < 7:
  print("Recuperação")
else:
  print("Reprovado")
print("Média:", media)
```

questão 14

```python
num_secreto = 7
tentativas = 0

while True:
  palpites = int(input("Tente descobrir o numero: "))
  tentativas += 1

  if palpites == num_secreto:
        print("Parabéns! Você acertou!")
        print("Número de tentativas: ", tentativas)
        break
  elif palpites < num_secreto:
        print("Tente um número maior")
  else:
        print("Tente um número menor")
```

questão 15

```python
todos_numeros = positivos = negativos = soma = 0

while True:
  numero = int(input("Digite um número inteiro (ou 0 para sair): "))

  if numero == 0:
    break
  todos_numeros += 1
  soma += numero
  if numero > 0:
    positivos += 1
  else:
    negativos += 1

print("Quantidade de números digitados: ", todos_numeros)
print("Quantidade de números positivos: ", positivos)
print("Quantidade de números negativos: ", negativos)
print("Soma dos números digitados: ", soma)

```
