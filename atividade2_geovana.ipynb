questão 13

Explicação: Mostra e conta todos os números primos entre dois valores A e B


 Entrada: A e B; Saída: lista de primos e quantidade.
Estruturas: Laços aninhados (um para percorrer o intervalo e outro para testar se cada número é primo)


Teste manual: Entre 10 e 15 -> 11 e 13 são primos. Total: 2.


```python
# Questão 13

a = int(input("digite o número (A): "))
b = int(input("digite o número (B, postivo e maior que A): "))
quantidade = 0
for numero in range (a, b + 1):
  primo = True
  for divisor in range(2, numero):
    if numero % divisor == 0:
      primo = False
      break
  if primo and numero > 1:
    print(numero)
    quantidade += 1
print("Quantidade de números primos: ", quantidade)
```

questão 14

Explicação: O usuário tenta adivinhar o número 37 com dicas de "maior" ou "menor" em até 5 chances


Entrada: palpites; Saída: dicas ou mensagem de vitória/derrota


Estruturas: Repetição for (5x) e break em caso de acerto


Teste manual: Chuta 30 -> "O secreto é maior". Chuta 37 -> "Acertou!".


```python
# Questão 14

secreto = 37

for tentativa in range (1,6):
    palpite = int(input(f"Tentativa {tentativa}/5 - Digite seu palpite: "))
    if palpite == secreto:
      print("Parbéns! Seu palpite estava certo!")
      break

    elif palpite < secreto:
      print("Dica: o número secreto é maior!")
    else:
      print("Dica: o número secreto é menor!")

else:
  print(f"Não foi dessa vez! Suas tentativas acabaram! O número era {secreto}")
```

questão 15

Explicação: Gerencia um caixa que inicia com R$ 200,00, processando entradas e saídas até que o código 0 seja digitado

saldo = valor disponível;

opção = se é entrada(1), retirada(2) ou encerramento(0);

valor = quantia que será movimentada;

q_entradas = contador de depósitos realizados;

q_retirada = contador de retiradas aprovadas;

q_recusada = contador das tentativas de retirada quando não há saldo;

total_entrada = acumulador que contabiliza o total que entrou no caixa;

total_saida = acumulador que contabiliza o total que saiu do caixa

Estruturas: Repetição while contínua e condições para validar o saldo

Teste manual:

Início: Saldo = 200.
Operação: Saque (2) de R$ 300.
Resultado: Exibe "Saldo insuficiente". O saldo permanece 200, contagem de recusadas = 1
.
Operação: Entrada (1) de R$ 50. Saldo = 250.



```python
# Questão 15

saldo = 200.00
q_entradas = q_retirada = q_recusada = 0
total_entrada = total_saida = 0.0

while True:
   print(f"Saldo atual: R${saldo:.2f}")
   tipo = int(input("Tipo (1: ENTRADA, 2: RETIRADA, 0: ENCERRAR):"))
   if tipo == 0:
    break
   if tipo == 1:
    valor = float(input("Valor do depósito: "))
    saldo += valor
    q_entradas += 1
    total_entrada += valor
   elif tipo == 2:
    valor = float(input("Valor do saque: "))
    if valor <= saldo:
      saldo -= valor
      q_retirada += 1
      total_saida += valor
      print("Valor retirado com sucesso!")
    else:
      print("Erro: Saldo insuficiente")
      q_recusada += 1
else:
    print("Opção inválida, tente novamente.")

print(f"Saldo final: ", saldo)
print(f"Total de entradas: R$ {total_entrada:.2f} ({q_entradas} operações)")
print(f"Total de saídas: R$ {total_saida:.2f} ({q_retirada} operações)")
print(f"Retiradas Recusadas: {q_recusada}")


```
