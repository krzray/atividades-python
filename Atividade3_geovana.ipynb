```python
nomes = []
quantidade = 0

while True:
    nome = input("Digite um nome (ou FIM para encerrar): ")
    if nome == "FIM":
        break
    novo = 1
    for n in nomes:
        if n == nome:
            novo = 0
    if novo == 1:
        nomes.append(nome)
        quantidade += 1

print("Lista de nomes cadastrados:", nomes)
print("Quantidade de nomes diferentes:", quantidade)
```

questão 12

```python

estoque = {"caramelo": 15, "babalu": 20, "amendoim": 10, "brownie": 30}

produto = input("Produto vendido: ")
quantos = int(input("Quantidade vendida: "))

if produto in estoque:
    if estoque[produto] >= quantos:
        estoque[produto] -= quantos
        print("Vendido com sucesso!")
    else:
        print("Estoque insuficiente!")
else:
    print("Produto não está cadastrado!")

print("Estoque final:", estoque)
```

```python
catalogo = []

for i in range(5):
    cod = int(input("Código: "))
    nome = input("Nome: ")
    preco = float(input("Preço: "))
    catalogo.append((cod, nome, preco))

procura = int(input("Digite o código para buscar: "))
existe = 0

for p in catalogo:
    if p[0] == procura:
        print("Nome:", p[1])
        print("Preço:", p[2])
        existe = 1

if existe == 0:
    print("Código não encontrado!")
```

questão 15

dicionário: representa cada item, permite organizar (codigo, nome, quantidade, preco)

lista: armazena o estoque completo, ela junta as infromações em sequencia, dá pra percorrer por meio do loop

tupla: como é imutável, é bom para guardar um resultado final (sem alterações)

```python
estoque = []

for i in range(5):
    item = {}
    item["codigo"] = int(input("Código: "))
    item["nome"] = input("Nome: ")
    item["quantidade"] = int(input("Quantidade: "))
    item["preco"] = float(input("Preço: "))
    estoque.append(item)

total_preco = 0
maior_valor = -1
nome_maior = ""

for item in estoque:
    valor_item = item["quantidade"] * item["preco"]
    total_preco += valor_item

    if valor_item > maior_valor:
        maior_valor = valor_item
        nome_maior = item["nome"]

tudo = (len(estoque), total_preco)

print("(a) Todos os itens:", estoque)
print("(b) Valor total em estoque:", total_preco)
print("(c) Item de maior valor acumulado:", nome_maior)
print("(d) Total geral (Qtd itens, Total):", tudo)
```
